```python
from sklearn.utils.class_weight import compute_class_weight
import numpy as np

classes = np.array([0, 1])
y = np.array([0, 0, 0, 1, 1, 0, 1, 0, 1, 0])  # Example labels

# Dict: just return the weights in order of classes
w_dict = compute_class_weight(class_weight={0: 5.0, 1: 2.0}, classes=classes, y=y)
print(w_dict)  # [5.0, 2.0]

# Balanced: compute based on counts
w_bal = compute_class_weight(class_weight="balanced", classes=classes, y=y)
print(w_bal)  # [0.833..., 1.25]  (assuming class 0 has 3 samples, class 1 has 2)
```
