```python
from pathlib import Path

workdir = Path('/eos/user/a/anunezde/Z_OUTPUT_eos/Run3_0626/NN_roster_final')
resultsdir = workdir / 'results'
```

```python
for file in resultsdir.iterdir():
    if file.is_file():
        filename = file.stem
        if 'hh_bbtautau' in filename:
            new_name = filename.replace("hh_bbtautau", "bbtautau") + file.suffix
            new_path = file.with_name(new_name)
            file.rename(new_path)
            print(f"Renaming: {file.name} → {new_name}")
        elif 'hh_bbww' in filename:
            new_name = filename.replace("hh_bbww", "bbww") + file.suffix
            new_path = file.with_name(new_name)
            file.rename(new_path)
            print(f"Renaming: {file.name} → {new_name}")
```
